```python
import pandas as pd

df = pd.read_csv("DSA Time complexities - Sheet1.csv")
df.head(27)
```

```python
df.hist()
```

```python
df.groupby("Algorithms")[['n','Execution Time (ms)']].sum()
```

```python
df.groupby("Algorithms")[['n','Execution Time (ms)']].mean().hist(bins=3)
```

```python
df.groupby("Algorithms")["Execution Time (ms)"].mean()
```

```python
df.groupby(["Algorithms", "n"])["Execution Time (ms)"].mean()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.lineplot(data =df, x='n', y='Execution Time (ms)', hue = 'Algorithms', marker = 'o')

plt.title("Sorting Algorithm Performance")
plt.xlabel("Dataset Size (N)")
plt.ylabel("Execution Time (ms)")
plt.show()
```

```python
df_filtered = df[~df["Algorithms"].str.contains("TopBoxOffice", case=False, na=False)]
```

```python
sns.lineplot(data =df_filtered, x='n', y='Execution Time (ms)', hue = 'Algorithms', marker = 'o')

plt.title("Sorting Algorithm Performance")
plt.xlabel("Dataset Size (N)")
plt.ylabel("Execution Time (ms)")
plt.show()
```
